```python
# function with parameters
def sample():
  print("Good Morning")
sample()
```

```python
# fuction without parameters
def sample(name):
  print("my name is:",name)
sample("john")
```

```python
# function with return type
def add(a,b):
  return a+b
print(add(2,4))
```

```python
# variables
# local variables
X = 10
def f1():
  print("X value from f1:", X)
def f2():
  print("X value from f2:", X)
f1()
f2()
print("X outside the function:",X)
```

```python
#global variable
X = 10
def f1():
  print("X value from f1:", X)
def f2():
  print("X value from f2:", X)
f1()
f2()
print("X value:", X)
```

```python
# arguments
def add(a,b):      # formal
  print(a+b)
add(2,4)           # actual
```

```python
# position arguments
def sample(a,b):
  print(a,b)
sample(2,3)
sample(3,2)
```

```python
# keyword arguments
def sample(a,b):
  print(a,b)
sample(a=2,b=3)
sample(b=3,a=2)
```

```python
def student_get_marks(result = "pass"):
  print("student result:",result)
student_get_marks()
student_get_marks("A+")
```

```python
# variablelength arguments
def name(*n):
  print(n)
name(1)
name(1,2)
name(1,2,3)
```

```python
def add(a, b):
    return a + b
def sub(a, b):
    return a - b
def mul(a, b):
    return a * b
def div(a, b):
    if b == 0:
        return "Error: Division by zero"
    return a / b
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")
choice = input("Enter choice (1/2/3/4): ")
if choice == "1":
    print("Result =", add(num1, num2))
elif choice == "2":
    print("Result =", sub(num1, num2))
elif choice == "3":
    print("Result =", mul(num1, num2))
elif choice == "4":
    print("Result =", div(num1, num2))
else:
    print("Invalid Choice")
```
